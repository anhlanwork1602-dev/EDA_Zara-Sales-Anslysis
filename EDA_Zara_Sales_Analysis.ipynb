# ### 4. DATA CLEANING



```python
df.isnull().sum()

```

```python
plt.figure(figsize=(12,5))

sns.heatmap(
    df.isnull(),
    cbar=False
)

plt.title("Missing Value")
plt.show()


```

```python
df.duplicated().sum()

```

```python
df.drop_duplicates(inplace=True)

```

```python
for col in df.columns:
    print(
        col, 
        df[col].nunique()
    )

```

# ### 5. CONVERT PRICE COLUMN



```python
df["price"].head()

```

# ### 6. CREATE NEW FEATURES



```python
df["price_category"]= pd.cut(
    df["price"],
    bins=[0,30,70,150,1000],
    labels=[
        "Low",
        "Medium",
        "High",
        "Luxury"
    ]
)

```

```python
df["sales_level"] = pd.qcut(
    df["Sales Volume"],
    q=4,
    labels=[
    "Low",
    "Medium",
    "High",
    "Very High"   
    ]
)

```

# ### 7. UNIVARIATE ANALYSIS



# a. Product category distribution
# Question: Zara sells which categories most?



```python
plt.Figure(figsize=(12,5))

sns.countplot(
    data=df, 
    y="Product Category",
    order=df["Product Category"].value_counts().index
)

plt.title(
    "Product Category Distribution"
)

plt.show()

```

# b. Sale distribution



```python
plt.figure(figsize=(10,5))

sns.histplot(
    df["Sales Volume"],
    bins=50,
    kde=True
)

plt.title(
    "Sales Volume Distribution"
)

plt.show()

```

# c. Price distribution



```python
plt.figure(figsize=(10,5))

sns.histplot(
    df["price"],
    bins=50,
    kde=True
)

plt.title(
    "Price Distribution"
)

plt.show()

```

# ### 8. PRODUCT POSITION ANALYSIS



# Research question: Products places where sell better?



```python
df["Product Position"].value_counts()

```

# Sales by position



```python
position_sales = (
    df.groupby(
        "Product Position"
    )["Sales Volume"]
    .mean()
    .sort_values()  
)

position_sales.plot(
    kind="bar",
    figsize=(10,5)
)

plt.title(
    "Average Sales by Product Position"
)

plt.show()

```

# ### 9. PROMOTION ANALYSIS



# Business question: Does promotion increase sales?



```python
sns.countplot(
    data=df,
    x="Promotion"
)

plt.show()

```

# Sales comparison



```python
sns.boxplot(
    data=df,
    x="Promotion",
    y="Sales Volume"
)

plt.title(
    "Sales vs Promotion"
)

plt.show()

```

# Mean difference



```python
df.groupby(
    "Promotion"
)["Sales Volume"].mean()

```

# ### 10. SEASONAL ANALYSIS



# Question: Are seasonal products easier to sell?



```python
sns.boxplot(
    data=df,
    x="Seasonal",
    y="Sales Volume"
)

plt.show()

```

# ### 11. CATEGORY VS SALES



# Very important. 



```python
category_sales = (
    df.groupby(
        "Product Category"
)["Sales Volume"]
.mean()
.sort_values(
    ascending=False
)
)

category_sales.plot(
    kind="bar",
    figsize=(12,5)
)

plt.title(
    "Average Sales by Category"
)

plt.show()

```

# ### 12. PRICE VS SALES



```python
sns.scatterplot(
    data=df,
    x="price",
    y="Sales Volume",
    alpha=0.4
)

plt.title(
    "Price vs Sales Volume"
)

plt.show()


```

# ### 13. CORRELATION ANALYSIS



```python
numeric = df.select_dtypes(
    include="number"
)

numeric.head()

```

```python
plt.figure(figsize=(10,6))

sns.heatmap(
    numeric.corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title(
    "Correlation Matrix"
)

plt.show()

```
